# Convert Parasound ACF/IDX to SEG-Y

Edit the paths in the next cell before running. The same input can be exported as `envelope`, `full_waveform`, or both.

```python
from pathlib import Path

from marine_sbp_asd2segy import convert_idx_outputs, convert_many, export_positions_csv, find_idx_files

idx_path = Path("path/to/line.asd.acf.idx")
raw_dir = Path("path/to/raw_directory")
output_dir = Path("outputs")
output_epsg = 32634  # example: WGS84 / UTM zone 34N
delay_ms = 0
trace_length_ms = 250
```

## Convert One IDX/ACF Pair

```python
if idx_path.exists():
    results = convert_idx_outputs(
        idx_path,
        output_dir=output_dir,
        output_epsg=output_epsg,
        outputs=("envelope", "full_waveform"),
        delay_ms=delay_ms,
        trace_length_ms=trace_length_ms,
    )
    for result in results:
        print(result)
else:
    print(f"Set idx_path to an existing .idx file: {idx_path}")
```

## Batch Convert a Directory

```python
if raw_dir.exists():
    idx_files = find_idx_files(raw_dir, recursive=True)
    print(f"Found {len(idx_files)} idx files")
    results = convert_many(
        idx_files,
        output_dir=output_dir,
        output_epsg=output_epsg,
        outputs=("envelope", "full_waveform"),
        delay_ms=delay_ms,
        trace_length_ms=trace_length_ms,
    )
    print(f"Wrote {len(results)} SEG-Y files")
else:
    print(f"Set raw_dir to an existing directory: {raw_dir}")
```

## Export Navigation Positions

```python
if raw_dir.exists():
    idx_files = find_idx_files(raw_dir, recursive=True)
    csv_path = export_positions_csv(idx_files, output_dir / "navigation.csv")
    print(csv_path)
else:
    print(f"Set raw_dir to an existing directory: {raw_dir}")
```
